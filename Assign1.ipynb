```python
''' 
Print Pyramid
n=4
'''
n=int(input("Enter the value of n: "))
str=1
spc=n-1
for i in range (n):
    for j in range (spc):
        print(" ",end=" ")
    for k in range (str):
        print('*', end=' ')
    print()
    spc -=1
    str +=2
```
